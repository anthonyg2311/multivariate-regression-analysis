```python
import numpy as np
import pandas as pd
```

```python
def normalize(data):
  for i in data.columns:
    data[i] = (data[i] - np.min(data[i])) / (np.max(data[i]) - np.min(data[i]))
  return data
```

```python
data = pd.read_csv("Advertising.csv")
data = normalize(data)
data
```

```python
X = data.drop(["Sales", data.columns[0]], axis = 1)
Y = data['Sales']
```

```python
def cost(thetaVec, b, X, Y):
  cost = 0
  for i in range(len(X)):
    cost += (1/len(X)) * ((np.dot(thetaVec, X.iloc[i]) +b) - Y.iloc[i])**2
  return cost
```

```python
def gc(thetaVec, b, lr, X, Y):
  #Initialization
  gradientVec = np.zeros(len(X.columns))
  db = 0

  #Finding Gradients
  for j in range(len(X.columns)): # column
    for i in range(len(X)): # row
      gradientVec[j] += (2/len(X)) * ((np.dot(thetaVec, X.iloc[i]) + b) -Y.iloc[i]) * X.iloc[i][j]
      db += (2/len(X)) * ((np.dot(thetaVec, X.iloc[i]) + b) -Y.iloc[i])

  #First step
  for j in range(len(gradientVec)):
    thetaVec[j] = thetaVec[j] - (lr * gradientVec[j])
  b = b - (lr * db)
  return thetaVec, b

#Parameter Initialization
thetaVec = np.zeros(len(X.columns))
b = 0
lr = 0.1
epochs = 100
for i in range(epochs):
  thetaVec, b = gc(thetaVec, b, lr, X, Y)

#Outputs
print("My Coefficients: ", thetaVec,"\nMy Intercept: ", b)
print("My error: ", cost(thetaVec, b, X, Y))
```

SkLearn Comparison

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Initialize and Fit
sklearn_model = LinearRegression()
sklearn_model.fit(X, Y)

# coefficients and intercept
print("Scikit-learn Coefficients:", sklearn_model.coef_)
print("Scikit-learn Intercept:", sklearn_model.intercept_)

# Make predictions using the scikit-learn model
sklearn_predictions = sklearn_model.predict(X)

sklearn_mse = cost(sklearn_model.coef_, sklearn_model.intercept_, X, Y)
print("Scikit-learn Mean Squared Error:", sklearn_mse)
```
