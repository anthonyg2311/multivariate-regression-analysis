```python
import numpy as np
import pandas as pd
```

```python
def normalize(data):
  for j in range(data.shape[1]):
    for i in range(len(data)):
      data.iloc[i, j] = (data.iloc[i, j] - min(data.iloc[:, j])) / (max(data.iloc[:, j]) - min(data.iloc[:, j]))
  return data
```

```python
data = pd.read_csv("Advertising.csv")
data = data.drop('Unnamed: 0', axis=1)
data
```

```python
normalize(data)
X = data.drop('Sales', axis = 1)
Y = data['Sales']
X= np.array(X)
```

```python
def costFunction(numExamples, thetaVec, b):
  cost = 0
  for i in range(numExamples):
    cost+=(1/numExamples) * ((np.dot(thetaVec, X[i]) + b) - Y[i]) **2
  return cost
```

```python
def stoch_gradientDescent(thetaVec, b, lr, numExamples):
  gradientVec = np.zeros(3)
  db = 0
  randIdx = np.random.randint(numExamples) #As what stochastic implied, which uses random values index from data set.

  for j in range(len(gradientVec)):
    gradientVec[j] = (2/numExamples) * ((np.dot(thetaVec, X[randIdx]) + b) - Y[randIdx]) * X[randIdx][j]
    db = (2/numExamples) * ((np.dot(thetaVec, X[randIdx]) + b) - Y[randIdx])

  for j in range(len(gradientVec)):
    thetaVec[j] = thetaVec[j] - (lr * gradientVec[j])
  b = b - (lr * db)
  return thetaVec, b

#Parameter initialization
thetaVec = np.zeros(3)
b = 0
lr = 0.9
numExamples = len(data)
epochs = 1000 #number of iterations

#Output
for i in range(epochs):
  thetaVec, b = stoch_gradientDescent(thetaVec, b, lr, numExamples)
print("Weights: ",thetaVec, b)
print("Error: ",costFunction(numExamples, thetaVec, b))
```

SkLearn Comparison

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, Y)
print("Weights: ", lin_reg.intercept_, lin_reg.coef_)
print("Error: ", costFunction(200, lin_reg.coef_, lin_reg.intercept_))
```
